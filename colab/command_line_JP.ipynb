```python
# @title 実行
# @markdown 文字起こし / 翻訳したいライブストリームの URL を入力してください（デフォルト値は例です）：
URL = "https://www.youtube.com/watch?v=r1dytEjbUqo" # @param {type:"string"}
model = "large" # @param ["small", "medium", "large", "turbo"] {allow-input: true}
language = "ja" # @param {type:"string"}
translation_prompt = "日本語から中国語に翻訳してください。テキストはリアルタイムの音声認識によるもので、認識エラーが含まれる可能性があります。翻訳時には文脈に応じて修正してください。" # @param {type:"string"}
translation_engine = "google" # @param ["openai", "google"]
api_key = "" # @param {type:"string"}
# @markdown （APIキーのスクレイピングと盗用が頻発しているため、試用APIキーは提供できません。ご自身のAPIキーをご記入ください。）
gpt_model = "gpt-5.6-luna" # @param ["gpt-5.4-nano", "gpt-5.4-mini", "gpt-5.6-luna", "gpt-5.6-terra", "deepseek-v4-flash", "kimi-k2-instruct", "glm-4.7-flash", "qwen3.5-flash"] {allow-input: true}
gemini_model = "gemini-3.5-flash-lite" # @param ["gemini-3.5-flash-lite", "gemini-3.5-flash", "gemini-3.6-flash"] {allow-input: true}
openai_base_url = "https://api.openai.com/v1" # @param ["https://api.openai.com/v1", "https://api.deepseek.com/v1", "https://api.moonshot.cn/v1", "https://open.bigmodel.cn/api/paas/v4", "https://dashscope.aliyuncs.com/compatible-mode/v1", "https://openrouter.ai/api/v1", ""] {allow-input: true}
google_base_url = "https://generativelanguage.googleapis.com" # @param ["https://generativelanguage.googleapis.com", ""] {allow-input: true}
# @markdown （DeepSeek、Moonshot、ZhipuAI、Alibaba などの OpenAI 互換ベンダーを使用するには：translation_engine を "openai" に設定し、openai_base_url でそのベンダーのベース URL を選択し、gpt_model をそのベンダーのモデル名に設定します。ベース URL ドロップダウンの最後の空オプションで任意のカスタムエンドポイントを入力できます。）
translation_history_size = 3 # @param {type:"slider", min:0, max:10, step:1}

!stream-translator-gpt "{URL}" --model {model} --language {language} \
  --use-faster-whisper \
  --translation-prompt "{translation_prompt}" \
  --{translation_engine}-api-key "{api_key}" \
  --gpt-model "{gpt_model}" --gemini-model "{gemini_model}" \
  --openai-base-url "{openai_base_url}" --google-base-url "{google_base_url}" \
  --translation-history-size {translation_history_size}
```
